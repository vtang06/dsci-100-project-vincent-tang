#### Task 3.1: Tidy the data (5 points)

- Create new variables (if needed)
- Transform the data into a tidy form (if needed)

```r
# Select the variables for research question
basketball_tidy <- basketball_data |>
    select(seed, reg_wins_pct, bid) |>

    # Remove rows with missing values
    filter(!is.na(seed), !is.na(reg_wins_pct), !is.na(bid)) |>

    # Convert bid into a categorical variable
    mutate(bid = as_factor(bid))

# Show the first 6 rows of the tidy dataset
head(basketball_tidy)
```

```r
# Distribution plot of regular-season win percentages
ggplot(basketball_tidy, aes(x = reg_wins_pct)) +
    geom_histogram(binwidth = 5, fill = "steelblue", color = "white") +
    labs(title = "Distribution of Regular-Season Win Percentages",
         x = "Regular-Season Win Percentage (%)",
         y = "Number of Teams")
```

```r
# Plotting relative proportions of automatic and at-large bids
ggplot(basketball_tidy, aes(x = bid)) +
    geom_bar(aes(y = after_stat(count / sum(count))),
             fill = "steelblue") +
    labs(title = "Proportion of NCAA Tournament Bid Types",
         x = "Tournament Bid Type",
         y = "Proportion of Teams")
```

For my data analysis, I plan to use K-Nearest Neighbours (KNN) classification to predict whether a team received an automatic or at-large bid to the NCAA Women's Basketball Tournament using tournament seed and regular-season win percentage. This method is appropriate because the outcome variable, bid type, is categorical with two possible classes. Specifically, the model will classify teams based on the bid types of teams with similar predictor values.
 
Potential limitations of KNN classification is that its performance depends on the number of neighbours selected. It is also sensitive to differences in the scales of the predictor variables, so the numeric predictors will need to be standardized. In addtion, tournament seed is assigned during tournament selection, therefore this model would be more useful for classifying historical bid types rather than predicting them before selection.

The way I plan to process the data is by randomly splitting the tidy dataset into 75% training data and 25% testing data using a fixed random seed for reproducibility. I will use 5-fold cross-validation on the training data to select the best number of neighbours (K). The numeric predictors will be centred and scaled using a recipe fitted only on the training data to prevent data leakage.

Lastly, I will evaluate the model using classification accuracy and a confusion matrix on the test set. Accuracy will show the proportion of correctly classified teams, while the confusion matrix will help identify which bid types the model tends to misclassify.
